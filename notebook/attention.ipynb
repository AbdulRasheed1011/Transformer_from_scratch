```python
import tensorflow as tf
from src.utils.config import load_config
import numpy as np
```

```python
q = np.float32(np.random.rand(4, 512, 512))
k = np.float32(np.random.rand(4,512, 512))
v = np.float32(np.random.rand(4,512, 512))
```

```python
def scaled_dot_product(q, k, v, mask = None):
    dk =  tf.cast(tf.shape(k)[-1], tf.float32)
    scores = tf.matmul(q, k, transpose_b = True)/tf.math.sqrt(dk)
    if mask is not None:
        scores += tf.cast(mask, tf.float32)*(-1e9)
    attn_weights = tf.nn.softmax(scores)
    out = tf.matmul(scores, v)

    return attn_weights, out

```

```python
def _split_heads(self, x):
        """
        x: (B, L, d_model)
        -> (B, H, L, Dh)
        """
        B = tf.shape(x)[0]
        L = tf.shape(x)[1]
        x = tf.reshape(x, (B, L, self.num_heads, self.depth))
        return tf.transpose(x, perm=[0, 2, 1, 3])
```
